```python
## step 2== Data cleaning 
apps_df = apps_df.dropna(subset=['Rating'])
for column in apps_df.columns:
    apps_df[column].fillna(apps_df[column].mode()[0],inplace=True)
apps_df.drop_duplicates(inplace= True)
apps_df = apps_df[apps_df['Rating'] <=5]
reviews_df.dropna(subset=['Translated_Review'], inplace= True)
    
```

```python
## Merge dataset on apps and handle non matching apps 
merged_df =  pd.merge(apps_df, reviews_df, on= 'App', how='inner')
```

```python
## step 3== data transformation 
apps_df['Reviews'] = apps_df['Reviews'].astype(int)
apps_df['Installs'] = apps_df['Installs'].str.replace(',','').str.replace('+', '').astype(int) 
apps_df['Price']  = apps_df['Price'].str.replace('$', '').astype(float)
```

```python
def convert_size(size):
    if 'M' in size:
        return float(size.replace('M', ''))
    elif 'k' in size:
        return float(size.replace('k', ''))/1024
    else:
        return np.nan
```

```python
apps_df['Size'] = apps_df['Size'].apply(convert_size)
```

```python
## Add log_installs and log_reviews columns:
apps_df['Log_Installs'] = np.log1p(apps_df['Installs'])
apps_df['Log_Reviews'] = np.log1p(apps_df['Reviews'])
```

```python
## Add rating group columns 
def rating_group(rating):
    if rating >= 4:
        return ' Top rated'
    elif rating >=3:
        return  'Above average'
    elif rating >=2:
        return ' Below average'

apps_df['Rating_group'] = apps_df['Rating'].apply(rating_group)
```

```python
## Add revenue columns 
apps_df['Revenue'] = apps_df['Price']* apps_df['Installs']
```

```python
## Sentiment Analysis 
sia = SentimentIntensityAnalyzer()
reviews_df['Sentiment_Score'] = reviews_df['Translated_Review'].apply(lambda x: sia.polarity_scores(str(x))['compound'])
```

```python
## Extract year from 'last updated and create year columns 
apps_df['Last Updated'] = pd.to_datetime(apps_df['Last Updated'], errors= 'coerce')
apps_df['Year'] = apps_df['Last Updated'].dt.year
```

```python
# ============================================================
# CREATE EXTRA COLUMNS REQUIRED FOR Q3, Q4 AND Q6
# ============================================================

# Android Version as numeric value
apps_df["Android_Version_Num"] = (
    apps_df["Android Ver"]
    .astype(str)
    .str.extract(r"(\d+(?:\.\d+)?)")[0]
    .astype(float)
)

# App name length
apps_df["App_Name_Length"] = (
    apps_df["App"]
    .astype(str)
    .str.len()
)

# Revenue
apps_df["Revenue"] = (
    apps_df["Price"] * apps_df["Installs"]
)

# First character of App
apps_df["App_Start"] = (
    apps_df["App"]
    .astype(str)
    .str[0]
    .str.upper()
)

# Whether App contains letter S
apps_df["App_Contains_S"] = (
    apps_df["App"]
    .astype(str)
    .str.contains("S", case=False, na=False)
)

# Whether App contains any number
apps_df["App_Contains_Number"] = (
    apps_df["App"]
    .astype(str)
    .str.contains(r"\d", regex=True, na=False)
)

print("Required columns created successfully.")
print([
    "Android_Version_Num",
    "App_Name_Length",
    "Revenue",
    "App_Start",
    "App_Contains_S",
    "App_Contains_Number"
])
```

```python
import plotly.express as px

# define the path for your Html files 
html_files_path = "./" 
os.makedirs(html_files_path, exist_ok=True)

plot_containers = ""

plot_width = 400
plot_height = 300
plot_bg_color = "black"
text_color = "white"
title_font = {"size":16}
axis_font = {"size": 12}


def save_plot_as_html(fig, filename, insight, time_id=None):
    global plot_containers

    filepath = os.path.join(html_files_path, filename)
    html_content =pio.to_html(
         fig,
         full_html= False,
         include_plotlyjs= "inline"
    )

    wrapper_id = time_id if time_id else filename.replace(".","_")

    plot_containers += f"""
    <div class="plot-container" id="{wrapper_id}"
         onclick="openPlot('{filename}')">
        <div class="plot">{html_content}</div>
        <div class="insights">{insight}</div>
        </div>
        """

    fig.write_html(
        filepath,
        full_html=False,
        include_plotlyjs="inline"
    ) 

def add_time_restricted_plot(fig, filename, insight, element_id):
    save_plot_as_html(
        fig,
        filename,
        insight,
        time_id=element_id
    )

## question=1: 
## Grouped bar: average rating + total review cout
## conditional:
## size >=10MB, January update, category avg rating >=4.0
## visible only 3 PM - 5PM IST
##===================================================================================

q1= apps_df[
    (apps_df["Size"]>=10)&
    (apps_df["Last Updated"].dt.month == 1)
    ].copy()

q1_grouped = (
    q1.groupby("Category", as_index= False)
      .agg(
          Total_Installs=("Installs", "sum"),
          Average_Rating=("Rating", "mean"),
          Total_Reviews=("Reviews", "sum")
      )
)

q1_grouped = q1_grouped[
    q1_grouped["Average_Rating"] >=4.0
    ].nlargest(10,"Total_Installs")

q1_long = q1_grouped.melt(
    id_vars=["Category"],
    value_vars=["Average_Rating", "Total_Reviews"],
    var_name="Metric",
    value_name="Value"
)

## Reviews are displayed in millions so both measures remain readable.

q1_long.loc[
    q1_long["Metric"] == "Total_Reviews", "Value"
    ] /= 1_000_000

q1_long["Matric"] = q1_long["Metric"].replace({
    "Average_Rating": "Average Rating",
    "Total_Reviews": "Total Reviews (Millions)"
})

fig_q1 = px.bar(
    q1_long,
    x="Category",
    y="Value",
    color="Metric",
    barmode="group",
    title="Q1:  Top 10 Categories - Rating vs Reviews",
    labels={"Value": "Value", "Category": "Category"},
    width=plot_width,
    height=plot_height
)
add_time_restricted_plot(
    fig_q1,
    "question1_grouped_bar.html",
    "Top 10 install categories after the January, size and average-rating filters.",
    "question1_time_chart"
)

# ============================================================
# ASSIGNMENT Q2
# Choropleth requirement
#
# IMPORTANT:
# The supplied Play Store dataset has no Country/Region column.
# Therefore a genuine global country choropleth cannot be created
# without inventing geographic information.
#
# We create a valid global choropleth only if a geographic column
# is present. Otherwise an explanatory Plotly figure is shown.
# Visible only 6 PM - 8 PM IST.
# ============================================================
q2_categories = (
    apps_df.groupby("Category", as_index=False)["Installs"]
    .sum()
    .nlargest(5, "Installs")
)

q2_categories = q2_categories[
    ~q2_categories["Category"].str.startswith(
        ("A", "C", "G", "S"),
        na=False
    )
]

# Detect an available geographic column without inventing values.
geo_candidates = [
    c for c in apps_df.columns
    if c.lower() in ["country", "countries", "region", "location"]
]

if geo_candidates:
    geo_col = geo_candidates[0]

    q2_geo = (
        apps_df[
            apps_df["Category"].isin(q2_categories["Category"]) &
            (~apps_df["Category"].str.startswith(
                ("A", "C", "G", "S"), na=False
            ))
        ]
        .groupby(geo_col, as_index=False)["Installs"]
        .sum()
    )

    fig_q2 = px.choropleth(
        q2_geo,
        locations=geo_col,
        locationmode="country names",
        color="Installs",
        color_continuous_scale="Plasma",
        title="Q2: Global Installs – Top Categories",
        width=plot_width,
        height=plot_height
    )
else:
    fig_q2 = go.Figure()
    fig_q2.add_annotation(
        text=(
            "Q2 requires a Country/Region column.<br>"
            "The supplied Play Store dataset does not contain geographic data."
        ),
        x=0.5,
        y=0.5,
        xref="paper",
        yref="paper",
        showarrow=False,
        font=dict(size=14, color="white")
    )
    fig_q2.update_layout(
        title="Q2: Global Installs Choropleth",
        xaxis=dict(visible=False),
        yaxis=dict(visible=False),
        width=plot_width,
        height=plot_height
    )

fig_q2.update_layout(
    paper_bgcolor=plot_bg_color,
    plot_bgcolor=plot_bg_color,
    font_color=text_color,
    margin=dict(l=10, r=10, t=45, b=10)
)

add_time_restricted_plot(
    fig_q2,
    "question2_choropleth.html",
    "Global category installs require geographic data in the source dataset.",
    "question2_time_chart"
)

# ============================================================
# ASSIGNMENT Q3
# Dual axis: average installs + revenue for free/paid
# Top 3 categories
# Conditions:
# installs >= 10k, revenue >= 10k, Android > 4.0,
# size > 15MB, Everyone, app name <= 30 chars
# Visible only 1 PM - 2 PM IST
# ============================================================
q3 = apps_df[
    (apps_df["Installs"] >= 10_000) &
    (apps_df["Revenue"] >= 10_000) &
    (apps_df["Android_Version_Num"] > 4.0) &
    (apps_df["Size"] > 15) &
    (apps_df["Content Rating"].eq("Everyone")) &
    (apps_df["App_Name_Length"] <= 30)
].copy()

top3_q3 = (
    q3.groupby("Category")["Installs"]
      .sum()
      .nlargest(3)
      .index
)

q3 = q3[q3["Category"].isin(top3_q3)].copy()

q3_grouped = (
    q3.groupby(["Category", "Type"], as_index=False)
      .agg(
          Average_Installs=("Installs", "mean"),
          Average_Revenue=("Revenue", "mean")
      )
)

fig_q3 = go.Figure()

for app_type in ["Free", "Paid"]:
    d = q3_grouped[q3_grouped["Type"] == app_type]

    fig_q3.add_trace(
        go.Bar(
            x=d["Category"],
            y=d["Average_Installs"],
            name=f"{app_type} - Avg Installs",
            offsetgroup=app_type,
            yaxis="y"
        )
    )

    fig_q3.add_trace(
        go.Scatter(
            x=d["Category"],
            y=d["Average_Revenue"],
            name=f"{app_type} - Avg Revenue",
            mode="lines+markers",
            yaxis="y2"
        )
    )

fig_q3.update_layout(
    title="Q3: Free vs Paid – Average Installs & Revenue",
    width=plot_width,
    height=plot_height,
    plot_bgcolor=plot_bg_color,
    paper_bgcolor=plot_bg_color,
    font_color=text_color,
    yaxis=dict(title="Average Installs"),
    yaxis2=dict(
        title="Average Revenue",
        overlaying="y",
        side="right"
    ),
    margin=dict(l=10, r=45, t=45, b=50)
)

add_time_restricted_plot(
    fig_q3,
    "question3_dual_axis.html",
    "Free and paid apps compared across the filtered top three categories.",
    "question3_time_chart"
)

# ============================================================
# ASSIGNMENT Q4
# Time series: total installs by category
# Conditions:
# app not start x/y/z, category starts E/C/B,
# reviews > 500, app name does not contain S
# Translation labels: Beauty-Hindi, Business-Tamil, Dating-German
# Visible only 6 PM - 9 PM IST
# ============================================================
q4 = apps_df[
    (~apps_df["App_Start"].isin(["X", "Y", "Z"])) &
    (apps_df["Category"].str.startswith(("E", "C", "B"), na=False)) &
    (apps_df["Reviews"] > 500) &
    (~apps_df["App_Contains_S"]) &
    (apps_df["Last Updated"].notna())
].copy()

q4["Month"] = q4["Last Updated"].dt.to_period("M").dt.to_timestamp()

q4_monthly = (
    q4.groupby(["Month", "Category"], as_index=False)["Installs"]
      .sum()
      .sort_values(["Category", "Month"])
)

q4_monthly["MoM_Growth"] = (
    q4_monthly.groupby("Category")["Installs"]
    .pct_change() * 100
)

q4_monthly["Translation"] = q4_monthly["Category"].map({
    "BEAUTY": "Hindi",
    "BUSINESS": "Tamil",
    "DATING": "German"
}).fillna("")

fig_q4 = px.line(
    q4_monthly,
    x="Month",
    y="Installs",
    color="Category",
    markers=True,
    title="Q4: Monthly Installs by Category",
    width=plot_width,
    height=plot_height
)

# Highlight months where MoM increase is >20%.
q4_spikes = q4_monthly[q4_monthly["MoM_Growth"] > 20]

for _, row in q4_spikes.iterrows():
    fig_q4.add_vrect(
        x0=row["Month"] - pd.Timedelta(days=10),
        x1=row["Month"] + pd.Timedelta(days=10),
        fillcolor="rgba(255,255,0,0.15)",
        line_width=0
    )

fig_q4.update_layout(
    plot_bgcolor=plot_bg_color,
    paper_bgcolor=plot_bg_color,
    font_color=text_color,
    margin=dict(l=10, r=10, t=45, b=50)
)

add_time_restricted_plot(
    fig_q4,
    "question4_time_series.html",
    "Yellow bands indicate category months where installs increased by more than 20% MoM.",
    "question4_time_chart"
)

# ============================================================
# ============================================================
# ASSIGNMENT Q5
# Bubble: size vs average rating
# Bubble size = installs
# Visible only 5 PM - 7 PM IST
# ============================================================

q5 = merged_df.copy()

# ------------------------------------------------------------
# 1. Rating → numeric
# ------------------------------------------------------------
q5["Rating_Q5"] = pd.to_numeric(
    q5["Rating"],
    errors="coerce"
)

# ------------------------------------------------------------
# 2. Reviews → numeric
# ------------------------------------------------------------
q5["Reviews_Q5"] = pd.to_numeric(
    q5["Reviews"]
    .astype(str)
    .str.replace(",", "", regex=False),
    errors="coerce"
)

# ------------------------------------------------------------
# 3. Installs → numeric
# ------------------------------------------------------------
q5["Installs_Q5"] = pd.to_numeric(
    q5["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False),
    errors="coerce"
)

# ------------------------------------------------------------
# 4. Size → numeric MB
# ------------------------------------------------------------
def convert_size_q5(value):

    value = str(value).strip().lower()

    if value in ["nan", "none", ""]:
        return np.nan

    try:

        if value.endswith("m"):
            return float(value[:-1])

        elif value.endswith("k"):
            return float(value[:-1]) / 1024

        else:
            return float(value)

    except:
        return np.nan


q5["Size_MB_Q5"] = q5["Size"].apply(convert_size_q5)

# ------------------------------------------------------------
# 5. Subjectivity
# ------------------------------------------------------------
if "Subjectivity" in q5.columns:

    q5["Subjectivity_Q5"] = pd.to_numeric(
        q5["Subjectivity"],
        errors="coerce"
    )

else:

    q5["Subjectivity_Q5"] = (
        q5["Translated_Review"]
        .fillna("")
        .astype(str)
        .apply(
            lambda x: TextBlob(x).sentiment.subjectivity
        )
    )

# ------------------------------------------------------------
# 6. Required categories
# ------------------------------------------------------------
allowed_q5_categories = [
    "GAME",
    "BEAUTY",
    "BUSINESS",
    "COMICS",
    "COMMUNICATION",
    "DATING",
    "ENTERTAINMENT",
    "SOCIAL",
    "EVENT"
]

# ------------------------------------------------------------
# 7. Apply Q5 conditions
# ------------------------------------------------------------
q5 = q5[
    (q5["Rating_Q5"] > 3.5) &

    (
        q5["Category"]
        .astype(str)
        .str.upper()
        .isin(allowed_q5_categories)
    ) &

    (q5["Reviews_Q5"] > 500) &

    (
        ~q5["App"]
        .astype(str)
        .str.contains(
            "S",
            case=False,
            na=False
        )
    ) &

    (q5["Subjectivity_Q5"] > 0.5) &

    (q5["Installs_Q5"] > 50_000)
].copy()

# ------------------------------------------------------------
# 8. Remove missing values
# ------------------------------------------------------------
q5 = q5.dropna(
    subset=[
        "Rating_Q5",
        "Reviews_Q5",
        "Installs_Q5",
        "Size_MB_Q5",
        "Subjectivity_Q5"
    ]
)


# ------------------------------------------------------------
# 9. FORCE NUMERIC COLUMNS BEFORE GROUPBY
# ------------------------------------------------------------

q5["Rating_Q5"] = pd.to_numeric(
    q5["Rating_Q5"],
    errors="coerce"
)

q5["Size_MB_Q5"] = pd.to_numeric(
    q5["Size_MB_Q5"],
    errors="coerce"
)

q5["Installs_Q5"] = pd.to_numeric(
    q5["Installs_Q5"],
    errors="coerce"
)

# Remove invalid rows
q5 = q5.dropna(
    subset=[
        "Rating_Q5",
        "Size_MB_Q5",
        "Installs_Q5"
    ]
).copy()

# Check datatypes
print("Rating:", q5["Rating_Q5"].dtype)
print("Size:", q5["Size_MB_Q5"].dtype)
print("Installs:", q5["Installs_Q5"].dtype)

# ------------------------------------------------------------
# 10. GROUP BY CATEGORY
# ------------------------------------------------------------

q5_grouped = (
    q5.groupby("Category", as_index=False)
      .agg(
          Average_Rating=("Rating_Q5", "mean"),
          Average_Size=("Size_MB_Q5", "mean"),
          Total_Installs=("Installs_Q5", "sum")
      )
)

print(q5_grouped)
# ------------------------------------------------------------
# CHECK DATA TYPES
# ------------------------------------------------------------
print("Q5 data types:")
print(q5_grouped.dtypes)

print("\nQ5 grouped data:")
print(q5_grouped)

# ------------------------------------------------------------
# 10. Bubble Chart
# ------------------------------------------------------------
fig_q5 = px.scatter(
    q5_grouped,
    x="Average_Size",
    y="Average_Rating",
    size="Total_Installs",
    color="Category",
    hover_name="Category",
    title="Q5: App Size vs Average Rating",
    labels={
        "Average_Size": "Average Size (MB)",
        "Average_Rating": "Average Rating",
        "Total_Installs": "Total Installs"
    },
    width=plot_width,
    height=plot_height
)

# ------------------------------------------------------------
# 11. Highlight Game in pink
# ------------------------------------------------------------
for trace in fig_q5.data:

    if str(trace.name).upper() == "GAME":
        trace.marker.color = "pink"

# ------------------------------------------------------------
# 12. Chart design
# ------------------------------------------------------------
fig_q5.update_layout(
    plot_bgcolor=plot_bg_color,
    paper_bgcolor=plot_bg_color,
    font_color=text_color,
    margin=dict(
        l=10,
        r=10,
        t=45,
        b=50
    )
)

# ------------------------------------------------------------
# 13. Time restriction
# ------------------------------------------------------------
add_time_restricted_plot(
    fig_q5,
    "question5_bubble.html",
    "Bubble size represents total installs; Game is highlighted in pink.",
    "question5_time_chart"
)
# ============================================================
# ASSIGNMENT Q6
# Stacked area: cumulative installs over time per category
# Conditions:
# rating >= 4.2, no numbers in app name,
# category starts T/P, reviews >1000,
# size 20-80 MB
# Translation legend:
# Travel & Local-French, Productivity-Spanish, Photography-Japanese
# Highlight months where category installs increase >25% MoM
# Visible only 4 PM - 6 PM IST
# ============================================================
q6 = apps_df[
    (apps_df["Rating"] >= 4.2) &
    (~apps_df["App_Contains_Number"]) &
    (apps_df["Category"].str.startswith(("T", "P"), na=False)) &
    (apps_df["Reviews"] > 1000) &
    (apps_df["Size"].between(20, 80, inclusive="both")) &
    (apps_df["Last Updated"].notna())
].copy()

q6["Month"] = q6["Last Updated"].dt.to_period("M").dt.to_timestamp()

q6_monthly = (
    q6.groupby(["Month", "Category"], as_index=False)["Installs"]
      .sum()
      .sort_values(["Category", "Month"])
)

q6_monthly["Cumulative_Installs"] = (
    q6_monthly.groupby("Category")["Installs"].cumsum()
)

q6_monthly["MoM_Growth"] = (
    q6_monthly.groupby("Category")["Installs"]
    .pct_change() * 100
)

translation_map_q6 = {
    "TRAVEL & LOCAL": "French",
    "PRODUCTIVITY": "Spanish",
    "PHOTOGRAPHY": "Japanese"
}

q6_monthly["Legend"] = q6_monthly["Category"].map(
    lambda x: f"{x} ({translation_map_q6.get(str(x).upper(), 'Original')})"
)

fig_q6 = px.area(
    q6_monthly,
    x="Month",
    y="Cumulative_Installs",
    color="Legend",
    title="Q6: Cumulative Installs by Category",
    labels={
        "Month": "Month",
        "Cumulative_Installs": "Cumulative Installs"
    },
    width=plot_width,
    height=plot_height
)

q6_spikes = q6_monthly[q6_monthly["MoM_Growth"] > 25]

for _, row in q6_spikes.iterrows():
    fig_q6.add_vrect(
        x0=row["Month"] - pd.Timedelta(days=10),
        x1=row["Month"] + pd.Timedelta(days=10),
        fillcolor="rgba(255,255,0,0.15)",
        line_width=0
    )

fig_q6.update_layout(
    plot_bgcolor=plot_bg_color,
    paper_bgcolor=plot_bg_color,
    font_color=text_color,
    margin=dict(l=10, r=10, t=45, b=50)
)

add_time_restricted_plot(
    fig_q6,
    "question6_stacked_area.html",
    "Yellow bands indicate category months where installs increased by more than 25% MoM.",
    "question6_time_chart"
)

# ============================================================
# FINAL HTML DASHBOARD
# ============================================================
dashboard_html = """
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">

<title>Google Play Store Reviews Analytics</title>

<style>
body {{
    font-family: Arial, sans-serif;
    background-color: #333;
    color: #fff;
    margin: 0;
    padding: 0;
}}

.header {{
    display: flex;
    align-items: center;
    justify-content: center;
    padding: 20px;
    background-color: #444;
    gap: 15px;
}}

.header img {{
    margin: 0 10px;
    height: 50px;
}}

.header h1 {{
    font-size: 26px;
}}

.container {{
    display: flex;
    flex-wrap: wrap;
    justify-content: center;
    padding: 20px;
}}

.plot-container {{
    border: 2px solid #555;
    margin: 10px;
    padding: 10px;
    width: {plot_width}px;
    height: {plot_height}px;
    overflow: hidden;
    position: relative;
    cursor: pointer;
    background-color: #333;
}}

.plot-container.time-hidden {{
    display: none !important;
}}

.insights {{
    display: none;
    position: absolute;
    right: 10px;
    top: 10px;
    max-width: 300px;
    background-color: rgba(0, 0, 0, 0.85);
    padding: 8px;
    border-radius: 5px;
    color: #fff;
    z-index: 100;
}}

.plot-container:hover .insights {{
    display: block;
}}

.time-message {{
    display: none;
    width: 100%;
    text-align: center;
    padding: 8px;
    color: #bbb;
    font-size: 13px;
}}
</style>

<script>
function openPlot(filename) {{
    window.open(filename, "_blank");
}}

/*
   Current IST time is checked in the browser.
   Therefore the dashboard automatically shows/hides
   each assignment chart while it remains open.
*/
function getISTMinutes() {{
    const now = new Date();

    const parts = new Intl.DateTimeFormat("en-US", {{
        timeZone: "Asia/Kolkata",
        hour: "2-digit",
        minute: "2-digit",
        hour12: false
    }}).formatToParts(now);

    let hour = parseInt(parts.find(p => p.type === "hour").value);
    const minute = parseInt(parts.find(p => p.type === "minute").value);

    // Some browsers can return 24 for midnight.
    if (hour === 24) hour = 0;

    return hour * 60 + minute;
}}

function isBetween(startHour, endHour) {{
    const minutes = getISTMinutes();
    return minutes >= startHour * 60 && minutes < endHour * 60;
}}

function updateRestrictedCharts() {{

    const rules = [
        ["question1_time_chart", 15, 17],
        ["question2_time_chart", 18, 20],
        ["question3_time_chart", 13, 14],
        ["question4_time_chart", 18, 21],
        ["question5_time_chart", 17, 19],
        ["question6_time_chart", 16, 18]
    ];

    rules.forEach(rule => {{
        const element = document.getElementById(rule[0]);

        if (!element) return;

        if (isBetween(rule[1], rule[2])) {{
            element.classList.remove("time-hidden");
        }} else {{
            element.classList.add("time-hidden");
        }}
    }});
}}

window.addEventListener("load", function() {{
    updateRestrictedCharts();

    // Recheck every 30 seconds while dashboard is open.
    setInterval(updateRestrictedCharts, 30000);
}});


</script>

</head>

<body>

<div class="header">
    <img
      src="https://upload.wikimedia.org/wikipedia/commons/thumb/4/4a/Logo_2013_Google.png/800px-Google.png"
      alt="Google Logo"
    >

    <h1>Google Play Store Reviews Analytics Dashboard</h1>

    <img
      src="https://upload.wikimedia.org/wikipedia/commons/thumb/7/78/Google_Play_Store_badge_EN.svg/1024px-Google_Play_Store_badge_EN.svg.png"
      alt="Google Play Store Logo"
    >
</div>

<div class="container">
    {plots}
</div>

</body>
</html>
"""

final_html = dashboard_html.format(
    plots=plot_containers,
    plot_width=plot_width,
    plot_height=plot_height
)


dashboard_path = os.path.join(
    html_files_path,
    "dashboard.html"
)

with open(dashboard_path, "w", encoding="utf-8") as f:
    f.write(final_html)

print("Dashboard created successfully:")
print(os.path.abspath(dashboard_path))

webbrowser.open(
    "file://" + os.path.realpath(dashboard_path)
)



    

    
    

```
